```python
df['uploads']=df['uploads'].fillna(df['uploads'].mean())
#detecting outliners with box plot
plt.boxplot(df.uploads,vert=False)
plt.xlabel('uploads')
plt.show()
```

```python
#calculate outlier boundaries and remove them
mean=df['uploads'].mean()
std=df['uploads'].std()
lowerbound=mean-2*std
upperbound=mean+2*std

df1=df[(df.uploads>=lowerbound) & (df.uploads<=upperbound)]

```

```python
df2=df1.fillna(0)
df2.isnull().sum()
```

```python
df2.duplicated().sum()
```

```python
#Univariate analysis
uploads=df2['uploads'].value_counts()
plt.figure(figsize=(6,8))
plt.bar(uploads,uploads.index,color='red')
plt.title("uploads by the youtubers")
plt.ylabel("uploads")
plt.xticks(np.arange(0,10,1))
plt.xlabel("youtuber")
plt.show()


```

```python
sns.set_style("darkgrid")
numerical_columns=df.select_dtypes(include=["int64","float64"]).columns
print(len(numerical_columns))
plt.figure(figsize=(14,len(numerical_columns)*3))

for idx,feature in enumerate(numerical_columns,1):
    plt.subplot(len(numerical_columns),2,idx)
    sns.histplot(df[feature],kde=True)
    plt.title(f"{feature}| Skewness:{round(df[feature].skew(),2)}")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,12))

sns.swarmplot(y='uploads',data=df2,color="skyblue")

plt.title("swarm of uploads")
plt.ylabel("uploads")
plt.show()
```
